```python
from google.colab import drive
import tensorflow as tf
from tensorflow.keras.preprocessing import image
import numpy as np
import os

drive.mount('/content/drive')

# PATHS TO YOUR SAVED MODELS
model_path = '/content/drive/MyDrive/Colab Notebooks/Chilli Plant Dataset/Saved_Models'

# Load the models
model_disease = tf.keras.models.load_model(f'{model_path}/chili_disease_mobilenet.h5')
model_growth = tf.keras.models.load_model(f'{model_path}/chili_growth_mobilenet.h5')

print("Models loaded successfully!")
```

```python
# Mappings must match the alphabetical order of your subfolders
disease_labels = {
    0: 'Bacterial Spot',
    1: 'Cercospora Leaf Spot',
    2: 'Curl Virus',
    3: 'Healthy Leaf',
    4: 'Nutrition Deficiency',
    5: 'White spot'
}

growth_labels = {
    0: 'Dry chili',
    1: 'Flower',
    2: 'Green Chili',
    3: 'Red Chili',
    4: 'Rotten Chili'
}
```

```python
def test_single_image(img_path):
    if not os.path.exists(img_path):
        print("Error: Image path does not exist.")
        return

    # 1. Load and Pre-process
    img = image.load_img(img_path, target_size=(224, 224))
    img_array = image.img_to_array(img) / 255.0  # Normalize to [0,1]
    img_array = np.expand_dims(img_array, axis=0) # Add batch dimension

    # 2. Run Predictions
    d_preds = model_disease.predict(img_array, verbose=0)
    g_preds = model_growth.predict(img_array, verbose=0)

    # 3. Get Results
    d_idx = np.argmax(d_preds)
    g_idx = np.argmax(g_preds)

    d_confidence = np.max(d_preds) * 100
    g_confidence = np.max(g_preds) * 100

    # 4. Display Output
    print("=" * 50)
    print(f"IMAGE: {os.path.basename(img_path)}")
    print("-" * 50)
    print(f"DISEASE DETECTION : {disease_labels[d_idx]} ({d_confidence:.2f}%)")
    print(f"GROWTH STAGE      : {growth_labels[g_idx]} ({g_confidence:.2f}%)")
    print("=" * 50)

# EXECUTE TEST
# Provide the path to a new image you want to test
my_test_image = "/content/Screenshot 2026-05-09 104925.png"
test_single_image(my_test_image)
```
